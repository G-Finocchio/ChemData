# Lasso up to 3ways test set performance (over 10 sets)

```python


results = []

# Define levels for 5-way
l1, l2, l3, l4, l5 = 6, 3, 11, 7, 3
levels_list = [l1, l2, l3, l4, l5]

# Example lambda grid for grid search ### One could try bigger grid searh to increase prediction performance but it takes already long enough since there are 10 splits
lambda_grid = [5e-4,1e-4,5e-5]#[5e-5, 1e-4, 5e-4, 1e-3, 5e-3]

range_main, range_theta, range_psi, range_phi, range_omega = get_ranges5(l1=l1, l2=l2, l3=l3, l4=l4, l5=l5)

# Loop through 10 random splits
for split_id in range(10):
    print(f"\n=== Running split {split_id} ===")

    # -----------------------------------------------------
    #  Load train/test data
    # -----------------------------------------------------
    df = pd.read_csv(f"../../../KMFM_main_paper/Data/Data_Suzuki_Miyaura/random_split_{split_id}.tsv.gz", sep="\t", compression="gzip")
    split = 4032  # fixed index boundary between train/test

    train_df = df.iloc[:split, :]
    test_df = df.iloc[split:, :]

    X_train = train_df.iloc[:, 1:-1].to_numpy()
    y_train = train_df.iloc[:, -1].to_numpy()
    X_test = test_df.iloc[:, 1:-1].to_numpy()
    y_test = test_df.iloc[:, -1].to_numpy()

    X_train=X_train[:,range_main+range_theta+range_psi]
    X_test=X_test[:,range_main+range_theta+range_psi]

    # -----------------------------------------------------
    #  Grid Search for best lambdas on training data
    # -----------------------------------------------------
    rez=grid_search_lasso(X_train, X_test, y_train, y_test,
        lambda_grid=lambda_grid,
        maxit=10, tol=0.5, sd_tol=1e-6, val_size=0.3, random_state=42,plot=True)
    best_lambda=rez[2]
    r2=rez[3]

    print(f"Split {split_id} R²: {r2:.2f}")

    # -----------------------------------------------------
    #  Save results
    # -----------------------------------------------------
    results.append({
        "split": split_id,
        "R2": r2,
        "lambda": best_lambda
    })

# -----------------------------------------------------
#  Summary across splits
# -----------------------------------------------------
results_df = pd.DataFrame(results)
print("\n=== Summary over splits ===")
print(results_df)
print(f"Average R²: {results_df['R2'].mean():.2f}")

# Optionally save summary to CSV
#results_df.to_csv("../../Results/Results_Suzuki_Miyaura/df_lasso_test_results.csv", index=False)

```
